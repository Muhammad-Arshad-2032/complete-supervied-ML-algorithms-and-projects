```python
### now to perform the vectorize opration we want to make the column of 1 
ones=np.ones([X.shape[0],1])
#
ones.shape
```

```python
X=X.reshape(X.shape[0],1)


#
X.shape
```

```python
## define the hypotesis
def Hypothesis(theta,X):
    return np.dot(theta,X)
```

```python
### define the cost function
def cost(Y,y_pred):
    m=len(Y)
    error=y_pred-Y
    return (1/(2*m))* np.sum(error**2)
#Gradient Descent
def gradient_descent(X, Y, theta, alpha, iterations):

    m = len(Y)

    for i in range(iterations):

        # Prediction
        y_pred = Hypothesis(X, theta)

        # Error
        error = y_pred - Y

        # Gradient
        gradient = (1 / m) * np.dot(X.T, error)

        # Update theta
        theta = theta - alpha * gradient

        # Print Cost
        if i % 100 == 0:
            print(f"Iteration {i}  Cost = {cost(Y, y_pred)}")

    return theta
```

```python
# Initial theta
theta = np.zeros(2)

# Hyperparameters
alpha = 0.1
iterations = 1000

# Training
theta = gradient_descent(X, Y, theta, alpha, iterations)

print("Theta =", theta)
#Prediction
y_pred = Hypothesis(X, theta)
print(y_pred)

```
