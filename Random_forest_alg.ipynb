```python
def Sample_row(df,percent):
    return df.sample(int(percent * df.shape[0]),replace= True)
```

```python
import random

def sample_col(df, percent):
    feature_cols = df.columns.to_list()[:-1]
    col = random.sample(
        feature_cols,
        int(percent * len(feature_cols))
    )
    return df[col]
```

```python
d1=Sample_row(df,0.1)
d2=Sample_row(df,0.8)
d3=Sample_row(df,0.7)
```

```python
from sklearn.tree import DecisionTreeClassifier
c1=DecisionTreeClassifier()
c2=DecisionTreeClassifier()
c3=DecisionTreeClassifier()
```

```python
c1.fit(d1.iloc[:,:-1],d1.iloc[:,:-1])
c2.fit(d2.iloc[:,:-1],d2.iloc[:,:-1])
c3.fit(d3.iloc[:,:-1],d3.iloc[:,:-1])
```
