```python
import sklearn # import neccssry library 
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd 
import numpy as np
# load the pretraind data set 
DataSet=load_breast_cancer()
DataSet


```

```python
NewDataSet = pd.DataFrame(DataSet.data, columns=DataSet.feature_names)
NewDataSet
NewDataSet['target']=DataSet.target
NewDataSet
```

```python
DataSet.target  # 0 means milignent and beigen
```

```python
DataSet.target_names # only two  target faeture are presnt in the dataset 
X=NewDataSet.iloc[:,:-1]
X



```

```python
Y=NewDataSet.iloc[:,-1] # for target feature 
Y
```

```python
NewDataSet.info()
NewDataSet.isnull()
NewDataSet.describe() # there is no null value i the dataset 
```

```python
# split the data 
results={}
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
models={
    'LogisticRegression':LogisticRegression(),
    'SVM':SVC(),
    'Decision Tree': DecisionTreeClassifier(),
}
for name, model in models.items():
       model.fit(X_train, Y_train)
       y_pred = model.predict(X_test)
       acc = accuracy_score(Y_test, y_pred)
       print(name, "Accuracy:", acc)
       results[name] = acc

```

```python
# Plot
import matplotlib.pyplot as plt 
plt.figure()
plt.bar(results.keys(), results.values())
plt.title("Model Comparison (Accuracy)")
plt.ylabel("Accuracy")
plt.xticks(rotation=45)
plt.show()
```

```python
import pickle
import streamlit as st
import numpy as np
import pickle

# load your trained model (important)
model = pickle.load(open("model.pkl", "rb"))

st.title("🧬 Breast Cancer Prediction App")

st.write("Enter 30 feature values below:")

# take inputs
inputs = []

for i in range(30):
    value = st.number_input(f"Feature {i+1}", step=0.01)
    inputs.append(value)

# predict button
if st.button("Predict"):
    input_data = np.array(inputs).reshape(1, -1)
    prediction = model.predict(input_data)

    if prediction[0] == 0:
        st.error("❌ Malignant (Cancer Detected)")
    else:
        st.success("✅ Benign (No Cancer)")
```
