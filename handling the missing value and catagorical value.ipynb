```python
import numpy as np 
import pandas as pd 
import sklearn 
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,OrdinalEncoder,StandardScaler
from sklearn.impute import SimpleImputer
```

```python
dataset=pd.read_csv("C:\\Users\\muham\\OneDrive\\Documents\\New folder\\employee_dataset.csv")
```

```python
dataset
```

### To check the missing  value

```python
dataset.isnull().sum()
```

```python
imputer=SimpleImputer(strategy='mean')
dataset[['Age','Salary','Experience']]=imputer.fit_transform(dataset[['Age','Salary','Experience']])
```

```python
dataset
## we can only achive this in pandas like
num_col=['Age','Salary','Experience']
for col in num_col:
    dataset[col]=dataset[col].fillna(dataset[col].mean())
dataset
for col in num_col:
    dataset[col]=dataset[col].fillna(dataset[col].mean())
dataset
```

### now we want to fill the catagorical values

```python

imputer=SimpleImputer(strategy='most_frequent')
dataset[['Gender','City']]=imputer.fit_transform(dataset[['Gender','City']])
# we can achive this in pandas 
num_col=['City','Gender']
for col in num_col:
    dataset[col]=dataset[col].fillna(dataset[col].mode())
dataset
```

```python
dataset
```

## now we want to encode the catagorical values

# oneHot encodig

```python
catgoric_column=['Gender','Department','Education','City']
oHE=OneHotEncoder()
encode=oHE.fit_transform(dataset[catgoric_column]).toarray()
encoded_dataset=pd.DataFrame(encode,
                            columns=oHE.get_feature_names_out(catgoric_column))
encoded_dataset
```
