```python
X=dataset.iloc[:,:1:] # retrive all the row and the only column 0, column 1 is not include 
X
X.shape
```

```python
Y=dataset.iloc[:,-1:] # rertrive all the row and column but the column -1 means last column not included
Y
```

## to split the data into the X & y train test

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

## Now Let's train the model linear regression

```python
model=LinearRegression()
```

## Now to fit the model using the train dataset(to learn the pattren)

```python
model.fit(X_train,Y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
mean=mean_squared_error(Y_test,y_pred)
mean
```
