```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
import warnings as war
war.filterwarnings("ignore")
```

```python
digit=load_digits()
```

```python
digit.data
digit.target_names
```

```python
X=digit.data
X=np.array(X)
Y=digit.target
Y=np.array(Y)
print(Y)
```

```python
digit.target_names
```

```python
digit.feature_names
```

```python
df=pd.DataFrame(X,columns=digit.feature_names)
df.head()
```

```python
df['target']=digit.target
df.head()
```

#### the data is split on the train and test

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3)
print("The train data is ")
print("The X_train  shape is ",X_train.shape)
print("The Y_train  shape is ",Y_train.shape)
print("The test data is ")
print("The X_test data is ",X_test.shape)
print("The Y_test data is ",Y_test.shape)
```

### fit the model

```python
model=LogisticRegression()
model.fit(X_train,Y_train)
```

### on tranining data

```python
X_train_prediction=model.predict(X_train)
print("model prediction on trainning data",X_train_prediction)
print("The actual traning data is ",Y_train)
```

### acuracy score for the training data

```python
accuracy_score(Y_train,X_train_prediction)*100
```

### confusion matrix for traning

```python
cm=confusion_matrix(Y_train,X_train_prediction)
cm
```

```python
sns.heatmap(cm,annot=True,fmt='d',cmap='coolwarm')
plt.xlabel("Model prediction on traning data")
plt.ylabel("Acual traning data")
plt.show()
```

#### on testing data

```python
X_test_prediction=model.predict(X_test)
print("model prediction on trainning data",X_test_prediction)
print("The actual traning data is ",Y_test)
```

### accuarcy on testing data 

```python
accuracy_score(X_test_prediction,Y_test)*100
```

### confusion matrix for testing

```python
cm1=confusion_matrix(Y_test,X_test_prediction)
cm
```

```python
sns.heatmap(cm1,annot=True,fmt='d',cmap='coolwarm')
plt.xlabel("Model prediction on testing data")
plt.ylabel("Acual testing data")
```
